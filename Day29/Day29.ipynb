HANDLING MISSING DATA:-----

```python
import numpy as np
import pandas as pd
```

CREATE DATASET

```python
data={
    'student_id':[1,2,3,4,5,6],
    'marks':[78,82,83,np.nan,90,95],
    'income':[20000,25000,32000,np.nan,46000,np.nan],
    'city':['delhi','mumbai',np.nan,'delhi','lucknow','delhi']
}
```

CONVERT DATA IN DATAFRAME

```python
df=pd.DataFrame(data)
```

NOW, CHECK THE DATA AND FIND THE TOTAL MISSING VALUES

```python
df.head(6)
```

```python
df.isnull().sum()
```

WE CAN ERASE ALL THE ROW THAT CONTAINS MISSING VALUES BUT NOT SUITABLE FOR SMALLER DATA:-----

```python
df_dropped=df.dropna()
print('DataFrame after dropping values')
df_dropped.head(6)
```

WE CAN USE STATISTICAL METHOD LIKE MEAN, MEADIAN, MODE OR ELSE:-----

HERE WE ALL TAKE THE PREVIOUS DATAFRAME BY COPYING IT

```python
df_new=df.copy()
```

```python
df_new.head(6)
```

```python
df_new['marks']=df_new['marks'].fillna(df_new['marks'].median())
df_new['income']=df_new['income'].fillna(df_new['income'].mean())
df_new['city']=df_new['city'].fillna(df_new['city'].mode()[0])
```
