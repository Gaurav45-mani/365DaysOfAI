HANDLING OUTLIERS:-----


```python
import numpy as np
lst_marks = [45, 32, 56, 75, 89, 54, 32, 89, 90, 87, 67, 54, 45, 98, 99, 67, 74, 1000, 1100]
```

```python
minimum,Q1,Q2,Q3,maximum=np.quantile(lst_marks,[0,0.25,0.5,0.75,1.0])
minimum,Q1,Q2,Q3,maximum
```

```python
IQR=Q3-Q1
IQR
```

CREATE FENCES, IF NUMBER LIE BETWEEN UPPER FENCE AND LOWER FENCE THEN IT WILL BE NOT A OUTLIERS

```python
lower_fence=Q1 - 1.5 * IQR
upper_fence=Q3 + 1.5 * IQR
```

```python
outliers=[]
non_outliers=[]
for i in lst_marks:
    if i>=lower_fence and i<=upper_fence:
        non_outliers.append(i)
    else:
        outliers.append(i)
print('OUTLIERS LIST:--',outliers)
print('NON OUTLIERS LIST:--',non_outliers)
```
