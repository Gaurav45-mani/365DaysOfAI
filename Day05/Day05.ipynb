IMPORT ALL REQUIRED LIBRARIES

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson
```

Assign lambda(Expected no. of event occur at a time interval) value 

```python
lam= 4
```

```python
x=list(range(11))
y=poisson.pmf(x,lam)
```

PLOT BARGRAPH

```python
plt.bar(x,y)
plt.title("Poisson Distribution")
plt.xlabel("Number of Events")
plt.ylabel("Probabilities")
plt.grid(alpha=0.1)
plt.show()
```
