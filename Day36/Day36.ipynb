Feature Engineering (Standardization)  :-----

```python
import seaborn as sns
import numpy as np
import pandas as pd
```

Loads the built-in restaurant tips dataset into a Pandas DataFrame.

```python
df = sns.load_dataset('tips')
df.head(1)
```

```python
total_bill = list(df['total_bill'])
```

```python
mean=np.mean(total_bill)
std=np.std(total_bill)
print(mean,std)
```

```python
standardized_data = []
for i in total_bill:
    z_score = (i - mean) / std
    standardized_data.append(z_score)
```

```python
new_standardized_data=pd.DataFrame(standardized_data,columns=['total_bill'])
```

```python
df_new=pd.concat([df,new_standardized_data],axis=1)
```
