Unit Vector Feature Scaling:----

```python
import seaborn as sns
import pandas as pd
from sklearn.preprocessing import normalize
```

Load IRIS Dataset

```python
df = sns.load_dataset('iris')
df.head(3)
```

```python
features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
scaled_array = normalize(df[features])
```

```python
new_scaled_array=pd.DataFrame(scaled_array,columns=['sepal_length', 'sepal_width', 'petal_length', 'petal_width'])
df_new=pd.concat([df,new_scaled_array],axis=1)
```
