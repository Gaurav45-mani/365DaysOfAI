# Label Encoding and Ordinal Encoding:----

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder
```

 Create a data and initialise the LabelEncoder:----
 - 1. Label Encoding -

```python
df_color = pd.DataFrame({'color': ['red', 'blue', 'green', 'green', 'red', 'blue']})
label_encoder = LabelEncoder()
color_encoded = label_encoder.fit_transform(df_color['color'])
color_encoded
```

   - 2. Ordinal Encoding -

```python
df_size = pd.DataFrame({'size': ['small', 'medium', 'large', 'medium', 'small', 'large']})
ordinal_encoder = OrdinalEncoder(categories=[['small', 'medium', 'large']])
size_encoded = ordinal_encoder.fit_transform(df_size[['size']])
size_encoded
```
