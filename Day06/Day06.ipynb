IMPORT ALL REQUIRED LIBRARIES

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
from scipy.stats import poisson
from scipy.stats import bernoulli
```

Put data in pandas DataFrame

```python
data=[45,48,55,56,67,58,98,67,99,83]
df=pd.DataFrame({"Data":data})
print(df)
```

Find all necessary descriptive statistics with these library

```python
print("Mean:",np.mean(df))
print("Median:",np.median(df))
print("Mode:",stats.mode(df,keepdims=False).mode)
print("Variance:",np.var(df))
print("Standard Deviation:",np.std(df))
```

Note----- we cannot find mode viabincount as numpy because data 
is  2d. So we can find it with pandas or stats 


Find quartile and IQR

```python
Q1=np.percentile(df,25)
Q2=np.percentile(df,50)
Q3=np.percentile(df,75)

IQR=Q3-Q1

print("Q1:",Q1)
print("Q2:",Q2)
print("Q3:",Q3)
print("IQR:",IQR)
```

Basic Visualization

```python
plt.hist(df,bins=5,edgecolor='black')
plt.title("MARKS HISTPLOT")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()

plt.boxplot(data)
plt.title("Marks")
plt.show()
```

COVARIANCE AND CORRELATION

```python
maths=[25,35,45,55,67,56,23,67]
science=[23,34,56,76,87,55,34,77]
df=pd.DataFrame({"maths_marks":maths,"science_marks":science})
print(df)
```

```python
print("Covariance:",df.cov())
print("Correlation:",df.corr())
print("Pearson_Correlation:",df.corr(method='pearson'))
print("Spearman_Correlation:",df.corr(method='spearman'))
      
      
```

Scatter plot

```python
plt.scatter(df['maths_marks'],df['science_marks'],edgecolor='red')
plt.title('Scatter plot between Maths and Science Marks')
plt.xlabel("Maths Marks")
plt.ylabel("Science Marks")
plt.grid(True)
plt.show()
```

Bernoulli distribution

```python
p=0.7
print("X=1 (Probability of success:",bernoulli.pmf(1,p))
print("X=0 (Probability of failure:",bernoulli.pmf(0,p))
```

```python
plt.bar([1,0],[bernoulli.pmf(1,p),bernoulli.pmf(0,p)])
plt.xticks([1,0],["Success","Failure"])
plt.ylabel('probability')
plt.title('Bernoulli distribution')
plt.show()
```

Poisson Distribution

```python
lam=0.7
x=list(range(11))
y=poisson.pmf(x,lam)
```

```python
plt.bar(x,y)
plt.title("Poisson Distribution")
plt.ylabel("Probabilities")
plt.grid(alpha=0.1)
plt.show()
```
