HANDLING IMBALANCED DATASET

```python
import numpy as np
import pandas as pd
from sklearn.utils import resample
```

```python
np.random.seed(123)
```

Sets total samples to 1000 with a 90% split for Class 0 (900 rows) and 10% for Class 1 (100 rows).

```python
n_samples = 1000
class_0_ratio = 0.9
n_class_0 = int(n_samples * class_0_ratio)       
n_class_1 = n_samples - n_class_0               
```

Builds the majority and minority class DataFrame

```python
class_0 = pd.DataFrame({
    'feature_1': np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature_2': np.random.normal(loc=0, scale=1, size=n_class_0),
    'target': [0] * n_class_0
})
class_1 = pd.DataFrame({
    'feature_1': np.random.normal(loc=2, scale=1, size=n_class_1),
    'feature_2': np.random.normal(loc=0, scale=1, size=n_class_1),
    'target': [1] * n_class_1
})
```

```python
df = pd.concat([class_0, class_1]).reset_index(drop=True)
```

Splits the combined dataset into two dedicated slices based on target labels.

```python
df_minority = df[df['target'] == 1]
df_majority = df[df['target'] == 0]
```

```python
df_minority_upsample = resample(
    df_minority, 
    replace=True,                 
    n_samples=len(df_majority),    
    random_state=42
)
df_upsampled = pd.concat([df_majority, df_minority_upsample]).reset_index(drop=True)
```

```python
df_majority_downsample = resample(
    df_majority,
    replace=False,                 
    n_samples=len(df_minority),    
    random_state=42
)
df_downsampled = pd.concat([df_majority_downsample, df_minority]).reset_index(drop=True)
```
