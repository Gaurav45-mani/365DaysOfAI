### Text Preprocessing & Null Handling
Missing entries (`NaN`) in text columns break string length and split calculations.

We apply `.fillna('')` to replace null values with empty strings, ensuring smooth feature extraction across all rows.

```python
df['Review Text'] = df['Review Text'].fillna(' ')
df['Title'] = df['Title'].fillna(' ')
```

### Structural Feature Extraction: Length & Volume
We extract total character count using `.str.len()` and word counts by splitting strings on spaces.

These features quantify customer feedback detail and review thoroughness.

```python
df['review_char_len'] = df['Review Text'].astype(str).str.len()
df['review_word_count'] = df['Review Text'].astype(str).apply(lambda x: len(x.split()))
```

### Vocabulary Complexity Metric Engineering
We synthesize `avg_word_length` by dividing character length by word count.

Using `np.where(df['review_word_count'] > 0, ..., 0)` prevents division-by-zero errors on empty review text entries.

```python
df['average_word_len'] = np.where(df['review_word_count'] > 0 ,df['review_char_len']/df['review_word_count'],0)
```

### Punctuation Feature Extraction
Punctuation marks like exclamation points (`!`) and question marks (`?`) indicate customer emotion and inquiry frequency.

We count these characters directly using `.apply(lambda x: x.count('!'))` to capture sentiment intensity signals.

```python
df['exclamation_count'] = df['Review Text'].astype(str).apply(lambda x : x.count('!'))
df['question_count'] = df['Review Text'].astype(str).apply(lambda x : x.count('?'))
```

```python
df['title_char_len'] = df['Title'].astype(str).str.len()
```
