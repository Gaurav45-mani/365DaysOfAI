# Target Guided Ordinal Encoding:----

```python
import pandas as pd
```

1. Create a sample DataFrame with a categorical feature ('City') and target ('price')

```python
data={
    'City':['New York', 'London', 'Paris', 'Tokyo', 'New York', 'Paris'],
    'price':[200, 150, 300, 250, 180, 320]
}
```

```python
df=pd.DataFrame(data)
df.head()
```

2. Calculate mean target price for each city

```python
mean_price=df.groupby('city')['price'].mean()
df['city_encoded'] = df['city'].map(mean_price)
```

```python
print("--- Target Guided Ordinal Encoded DataFrame ---")
df
```
